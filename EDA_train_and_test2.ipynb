# Exploratory Data Analysis (EDA)

Dataset: train_and_test2.csv

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8,5)
```

```python

df = pd.read_csv(r"C:\Users\msi-1\Downloads\archive (12)\train_and_test2.csv")
df.head()
```

## Basic Data Understanding

```python

df.shape
df.info()
df.describe()
```

## Missing Values Analysis

```python

df.isnull().sum()
```

```python

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.show()
```

## Univariate Analysis

```python

sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python

sns.boxplot(x=df['Fare'])
plt.title("Fare Boxplot")
plt.show()
```

```python

sns.countplot(x='Sex', data=df)
plt.title("Gender Count")
plt.show()
```

## Bivariate Analysis

```python

sns.scatterplot(x='Age', y='Fare', data=df)
plt.title("Age vs Fare")
plt.show()
```

```python

sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Passenger Class")
plt.show()
```

```python

sns.boxplot(x='Sex', y='Age', data=df)
plt.title("Age Distribution by Gender")
plt.show()
```

```python

sns.violinplot(x='Sex', y='Age', data=df)
plt.title("Age Distribution by Gender (Violin)")
plt.show()
```

## Multivariate Analysis

```python

sns.pairplot(df[['Age','Fare','Pclass','Survived']], hue='Survived')
plt.show()
```

## Correlation Analysis

```python

corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## Distribution Comparison

```python

sns.kdeplot(df[df['Survived']==1]['Age'], label='Survived')
sns.kdeplot(df[df['Survived']==0]['Age'], label='Not Survived')
plt.legend()
plt.title("Age Distribution by Survival")
plt.show()
```

## Advanced Plots

```python

sns.swarmplot(x='Pclass', y='Age', data=df)
plt.title("Swarm Plot: Age vs Pclass")
plt.show()
```

```python

sns.stripplot(x='Sex', y='Fare', data=df, jitter=True)
plt.title("Fare Distribution by Gender")
plt.show()
```
