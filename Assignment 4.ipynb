# Exploratory Data Analysis (EDA) using Sample Superstore Dataset

## Objective

The objective of this assignment is to perform Exploratory Data Analysis (EDA) on the Sample Superstore dataset using Pandas and Seaborn. Various statistical and graphical techniques are used to understand patterns, trends, and relationships in the data.

## Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset

```python
df = pd.read_csv(r"C:\Users\msi-1\Downloads\SampleSuperstore.csv")
df.head()
```

## Dataset Overview

```python
df.info()
```

```python
df.describe()
```

## Checking Missing Values

```python
df.isnull().sum()
```

## Data Cleaning

Removing duplicate records if any.

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

## Univariate Analysis

### Histogram – Sales Distribution

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Sales'], bins=20, kde=True)
plt.title('Sales Distribution')
plt.show()
```

### Count Plot – Category

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Category', data=df)
plt.title('Count of Products by Category')
plt.show()
```

## Bivariate Analysis

### Scatter Plot – Sales vs Profit

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Sales', y='Profit', data=df)
plt.title('Sales vs Profit')
plt.show()
```

### Box Plot – Profit by Category

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Category', y='Profit', data=df)
plt.title('Profit Distribution by Category')
plt.show()
```

## Multivariate Analysis

### Heatmap – Correlation Matrix

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['Sales','Profit','Discount','Quantity']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

### Pair Plot

```python
sns.pairplot(df[['Sales','Profit','Discount','Quantity']])
plt.show()
```

## Outlier Detection

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Sales'])
plt.title('Outliers in Sales')
plt.show()
```
